```python
from collections import deque

# =========================================
# Breadth First Search (BFS) -- MODIFIED DISPLAY LIST
# =========================================
def bfs(graph, start):
    # Set untuk menyimpan simpul yang sudah dikunjungi
    visited = set()

    # Queue digunakan untuk proses BFS
    # deque lebih efisien dibanding list untuk operasi antrian
    queue = deque([start])

    #Membuat list untuk menyimpan hasil transversal
    result = []

    # Selama queue masih memiliki isi
    while queue:
        # Mengambil simpul paling depan dari queue
        vertex = queue.popleft()

        # Jika simpul belum pernah dikunjungi
        if vertex not in visited:

            # Menandai simpul sebagai sudah dikunjungi
            visited.add(vertex)

            #Menambahkan simpul ke list result
            result.append(vertex)

            # Menambahkan semua tetangga yang belum dikunjungi
            # ke bagian belakang queue
            queue.extend([
                neighbor
                for neighbor in graph[vertex]
                if neighbor not in visited
            ])

    return result

#Contoh pemanggilan
bfs(graph, 'A')
```

```python
# =========================================
# Depth First Search (DFS)
# =========================================
def dfs(graph, start, visited=None):
    # Jika visited belum ada,
    # buat set kosong untuk menyimpan simpul yang sudah dikunjungi
    if visited is None:
        visited = set()

    # Jika simpul belum pernah dikunjungi
    if start not in visited:
        # Menampilkan simpul yang sedang dikunjungi
        print(start, end=" ")

        # Menandai simpul sebagai sudah dikunjungi
        visited.add(start)

        # Melakukan iterasi ke seluruh tetangga simpul
        for neighbor in graph[start]:
            # Memanggil DFS secara rekursif
            # untuk menelusuri simpul tetangga
            dfs(graph, neighbor, visited)

#Contoh pemanggilan
dfs(graph, 'A')
```

```python
# =========================================
# Menghitung jumlah tetangga setiap simpul
# =========================================
def count_neighbors(graph):
    # Dictionary untuk menyimpan hasil
    # Format:
    # {simpul: jumlah_tetangga}
    result = {}

    # Melakukan iterasi ke setiap simpul pada graph
    for node in graph:
        # Menghitung jumlah tetangga simpul
        # menggunakan len() pada list tetangga
        count = len(graph[node])

        # Menyimpan hasil ke dictionary result
        result[node] = count

    # Mengembalikan hasil akhir
    return result

print(count_neighbors(graph))
```

```python
# Graph tak berarah baru
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['A', 'C', 'E'],
    'E': ['C', 'D']
}
```

```python
# =========================================
# Mencari satu jalur dari start ke end
# menggunakan DFS
# =========================================
def find_path(graph, start, end, path=None):

    # Jika path belum ada,
    # buat list kosong untuk menyimpan jalur
    if path is None:
        path = []

    # Menambahkan simpul sekarang ke jalur
    path = path + [start]

    # Jika simpul tujuan ditemukan,
    # kembalikan jalurnya
    if start == end:
        return path

    # Jika simpul tidak ada dalam graph
    if start not in graph:
        return None

    # Menelusuri seluruh tetangga simpul
    for neighbor in graph[start]:

        # Jika tetangga belum ada di jalur
        # agar tidak terjadi loop tak hingga
        if neighbor not in path:

            # Mencari jalur secara rekursif
            new_path = find_path(graph, neighbor, end, path)

            # Jika jalur ditemukan,
            # kembalikan hasilnya
            if new_path:
                return new_path

    # Jika tidak ada jalur ditemukan
    return None

#Contoh pemanggilan
print(find_path(graph, 'A', 'D'))
```

```python
# =========================================
# Mengecek apakah graph terhubung
# menggunakan DFS
# =========================================

def is_connected(graph):

    # Set untuk menyimpan simpul
    # yang sudah dikunjungi
    visited = set()

    # Fungsi DFS rekursif
    def dfs(node):

        # Menandai simpul sebagai sudah dikunjungi
        visited.add(node)

        # Menelusuri seluruh tetangga simpul
        for neighbor in graph[node]:

            # Jika tetangga belum dikunjungi
            if neighbor not in visited:

                # Lanjut DFS ke tetangga
                dfs(neighbor)

    # Mengambil satu simpul pertama
    # sebagai titik awal DFS
    start_node = next(iter(graph))

    # Menjalankan DFS dari simpul awal
    dfs(start_node)

    # Graph dikatakan terhubung
    # jika semua simpul berhasil dikunjungi
    return len(visited) == len(graph)

#Pemanggilan
is_connected(graph)
```
