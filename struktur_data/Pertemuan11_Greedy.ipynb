```python
def coin_change_greedy(amount, coins):
  # Mengurutkan daftar koin dari nilai terbesar ke terkecil
  coins.sort(reverse=True)

  # Menyimpan koin-koin yang dipilih
  result = []

  # Memeriksa setiap jenis koin
  for coin in coins:
    # Selama nilai amount masih cukup untuk koin tersebut
    while amount >= coin:
      # Kurangi amount dengan nilai koin
      amount -= coin

      # Simpan koin yang digunakan ke hasil
      result.append(coin)

  # Mengembalikan daftar koin yang digunakan
  return result

# Contoh penggunaan
amount = 57              # Jumlah uang yang akan ditukar
coins = [25, 10, 5, 1]   # Denominasi koin yang tersedia

# Memanggil fungsi greedy untuk mencari kombinasi koin
change = coin_change_greedy(amount, coins)

# Menampilkan hasil koin yang digunakan
print("Koin yang digunakan:", change)
```

```python
def fractional_knapsack(items, capacity):
  # Mengurutkan item berdasarkan rasio nilai/berat (value/weight) dari terbesar ke terkecil
  # x[0] = value, x[1] = weight
  items = sorted(items, key=lambda x: x[0]/x[1], reverse=True)

  # Menyimpan total nilai maksimum yang berhasil dimasukkan ke knapsack
  total_value = 0.0

  # Memproses setiap item yang sudah diurutkan
  for value, weight in items:

    # Jika kapasitas masih cukup untuk mengambil seluruh item
    if capacity >= weight:

      # Kurangi kapasitas yang tersisa
      capacity -= weight

      # Tambahkan seluruh nilai item ke total nilai
      total_value += value

    # Jika kapasitas tidak cukup untuk mengambil seluruh item
    else:

      # Ambil sebagian item sesuai kapasitas yang tersisa
      # Nilai yang diperoleh proporsional terhadap bagian yang diambil
      total_value += value * (capacity / weight)

      # Knapsack sudah penuh, hentikan proses
      break

  # Mengembalikan total nilai maksimum yang diperoleh
  return total_value


# Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)] # (nilai, berat)

# Kapasitas maksimum knapsack
capacity = 50

# Memanggil fungsi untuk menghitung nilai maksimum
max_value = fractional_knapsack(items, capacity)

# Menampilkan hasil
print("Nilai maksimum yang dapat dibawa:", max_value)
```

```python
import heapq

def prim_mst(graph, start):
  # Menyimpan simpul yang sudah masuk ke MST
  visited = set()

  # Priority Queue (Min Heap)
  # Format: (bobot_sisi, simpul_tujuan)
  # Dimulai dari simpul awal dengan bobot 0
  min_heap = [(0, start)]

  # Menyimpan total bobot MST
  total_weight = 0

  # Selama masih ada simpul yang bisa diproses
  while min_heap:

    # Mengambil sisi dengan bobot terkecil dari heap
    weight, node = heapq.heappop(min_heap)

    # Jika simpul sudah dikunjungi, lewati
    if node in visited:
      continue

    # Tandai simpul sebagai sudah dikunjungi
    visited.add(node)

    # Tambahkan bobot sisi ke total MST
    total_weight += weight

    # Periksa semua tetangga dari simpul saat ini
    for neighbor, edge_weight in graph[node]:

      # Jika tetangga belum masuk MST
      if neighbor not in visited:

        # Masukkan sisi ke min heap
        heapq.heappush(min_heap, (edge_weight, neighbor))

  # Mengembalikan total bobot MST
  return total_weight


# Representasi graf menggunakan adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}

# Menjalankan algoritma Prim mulai dari simpul A
print("Total bobot MST (Prim):", prim_mst(graph, 'A'))
```

```python
def kruskal_mst(edges, n_nodes):
  # Membuat parent awal untuk setiap node (Union-Find)
  parent = {i: i for i in range(n_nodes)}

  def find(x):
    # Mencari root (induk utama) dari node x
    while parent[x] != x:
      x = parent[x]
    return x

  def union(x, y):
    # Mencari root dari kedua node
    root_x = find(x)
    root_y = find(y)

    # Jika root berbeda, gabungkan kedua himpunan
    if root_x != root_y:
      parent[root_y] = root_x
      return True

    # Jika root sama, berarti akan membentuk siklus
    return False

  # Mengurutkan sisi berdasarkan bobot terkecil
  edges.sort(key=lambda x: x[2])

  # Menyimpan total bobot MST
  total_weight = 0

  # Memproses setiap sisi yang sudah diurutkan
  for u, v, weight in edges:

    # Tambahkan sisi jika tidak membentuk siklus
    if union(u, v):
      total_weight += weight

  # Mengembalikan total bobot MST
  return total_weight


# edges: (node1, node2, weight)
# node direpresentasikan dengan bilangan integer
edges = [
  (0, 1, 2),
  (0, 2, 3),
  (1, 2, 1),
  (1, 3, 1),
  (2, 3, 4)
]

# Menjalankan algoritma Kruskal untuk 4 node
print("Total bobot MST (Kruskal):", kruskal_mst(edges, 4))
```
