```python
# Membuat class bernama SimpleHashTable
class SimpleHashTable:

  # Constructor untuk inisialisasi ukuran hash table
  def __init__(self, size):
    self.size = size                # Menyimpan ukuran table
    self.table = [None] * size      # Membuat list kosong sesuai ukuran

  # Fungsi hash untuk menentukan index penyimpanan
  def hash_function(self, key):
    return hash(key) % self.size    # Mengubah key menjadi index

  # Fungsi untuk menambahkan data ke hash table
  def insert(self, key, value):
    index = self.hash_function(key) # Mencari index berdasarkan key
    self.table[index] = value       # Menyimpan value pada index tersebut

  # Fungsi untuk mencari data berdasarkan key
  def search(self, key):
    index = self.hash_function(key) # Mencari index berdasarkan key
    return self.table[index]        # Mengembalikan value dari index

# Contoh penggunaan hash table
ht = SimpleHashTable(10)            # Membuat hash table ukuran 10

ht.insert("apel", 100)              # Menyimpan data apel = 100
ht.insert("pisang", 200)            # Menyimpan data pisang = 200

# Menampilkan nilai dari key "apel"
print('Nilai untuk "Apel"', ht.search('apel'))

# Menampilkan nilai dari key "pisang"
print('Nilai untuk "Pisang"', ht.search('pisang'))
```

```python
# Membuat class hash table dengan metode linear probing
class LinearProbingHashTable:

  # Constructor untuk inisialisasi ukuran hash table
  def __init__(self, size):
    self.size = size                # Menyimpan ukuran table
    self.table = [None] * size      # Membuat list kosong sesuai ukuran

  # Fungsi hash untuk menentukan index berdasarkan key
  def hash_function(self, key):
    return hash(key) % self.size    # Menghasilkan index dari key

  # Fungsi untuk memasukkan data ke hash table
  def insert(self, key, value):
    index = self.hash_function(key) # Menentukan index awal
    original_index = index          # Menyimpan index awal untuk pengecekan penuh

    # Melakukan probing jika slot sudah terisi
    while self.table[index] is not None:

      # Jika key sudah ada, hentikan perulangan
      if self.table[index][0] == key:
        break

      # Pindah ke index berikutnya
      index = (index + 1) % self.size

      # Jika kembali ke index awal berarti table penuh
      if index == original_index:
        raise Exception('Hash table penuh')

    # Menyimpan pasangan key dan value
    self.table[index] = (key, value)

  # Fungsi untuk mencari data berdasarkan key
  def search(self, key):
    index = self.hash_function(key) # Menentukan index awal
    original_index = index          # Menyimpan index awal

    # Melakukan pencarian selama slot tidak kosong
    while self.table[index] is not None:

      # Jika key ditemukan, kembalikan value
      if self.table[index][0] == key:
        return self.table[index][1]

      # Pindah ke index berikutnya
      index = (index + 1) % self.size

      # Jika kembali ke index awal, hentikan pencarian
      if index == original_index:
        break

    # Mengembalikan None jika data tidak ditemukan
    return None

# Contoh penggunaan
ht = LinearProbingHashTable(5)  # Membuat hash table ukuran 5

ht.insert('apel', 100)          # Menyimpan data apel = 100
ht.insert('pisang', 200)        # Menyimpan data pisang = 200
ht.insert('melon', 300)         # Menyimpan data melon = 300

# Menampilkan nilai dari key "apel"
print('Nilai untuk "apel"', ht.search('apel'))

# Menampilkan nilai dari key "melon"
print('Nilai untuk "melon"', ht.search('melon'))
```

```python
# Membuat class hash table dengan metode chaining
class ChainingHashTable:

  # Constructor untuk inisialisasi ukuran hash table
  def __init__(self, size):
    self.size = size                # Menyimpan ukuran table

    # Membuat list berisi list kosong untuk chaining
    self.table = [[] for _ in range(size)]

  # Fungsi hash untuk menentukan index berdasarkan key
  def hash_function(self, key):
    return hash(key) % self.size    # Menghasilkan index

  # Fungsi untuk memasukkan data ke hash table
  def insert(self, key, value):
    index = self.hash_function(key) # Menentukan index

    # Mengecek apakah key sudah ada pada bucket
    for i, (k, v) in enumerate(self.table[index]):

      # Jika key ditemukan, update value
      if k == key:
        self.table[index][i] = (key, value)
        return

    # Jika key belum ada, tambahkan ke bucket
    self.table[index].append((key, value))

  # Fungsi untuk mencari data berdasarkan key
  def search(self, key):
    index = self.hash_function(key)  # Menentukan index

    # Menelusuri semua data pada bucket
    for k, v in self.table[index]:

      # Jika key ditemukan, kembalikan value
      if k == key:
        return v

    # Mengembalikan None jika data tidak ditemukan
    return None

# Contoh penggunaan
ht = ChainingHashTable(5)     # Membuat hash table ukuran 5

ht.insert('apel', 100)        # Menyimpan data apel = 100
ht.insert('pisang', 200)      # Menyimpan data pisang = 200
ht.insert('melon', 300)       # Menyimpan data melon = 300

# Menampilkan nilai dari key "apel"
print("Nilai untuk 'apel':", ht.search('apel'))

# Menampilkan nilai dari key "melon"
print("Nilai untuk 'melon':", ht.search('melon'))
```

```python
class LinearProbingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [None] * size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        break
      index = (index + 1) % self.size
      if index == original_index:
        raise Exception('Hash table penuh')

    self.table[index] = (key, value)

  def search(self, key):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        return self.table[index][1]
      index = (index + 1) % self.size
      if index == original_index:
        break
    return None

# Uji
ht = LinearProbingHashTable(5)

# Inputan
input = {
    'A': 10,
    'B': 20,
    'C': 30,
    'D': 40,
    'E': 50
}

# Masukkan data
for k, v in input.items():
  ht.insert(k, v)

# Menampilkan isi tabel
print(ht.table)
```
